```python
a = 2; b = 3
print(a / b ) # True division	Quotient of a and b
print(a // b) # Floor division	Quotient of a and b, removing fractional parts
print(a % b ) # Modulus	Integer remainder after division of a by b
print(a ** b) # Exponentiation	a raised to the power of b
print(-a    ) # Negation	The negative of a
print(+a    ) # Unary plus	a unchanged (rarely used)
```

```python
x = [1, 2, 3]
print(type(x))
# Only good on floats: x.is_integer()
y = x
print(y, x)
x.append(4) # append 4 to the list pointed to by x
print(y, x) # y's list is modified as well!
x = 'something else'
print(y, x)  # y is unchanged
x = 4.5
print(x.is_integer())
print(type(x.is_integer))
print(x.real, "+", x.imag, 'i')
print(type(x))
y = x
x += 5  # add 5 to x's value, and assign it to x
print("x =", x)
print("y =", y)
x = 'hello'
type(x)
```
